```python
import numpy as np
import math
```

```python
R_C_A = np.array([[0, 0, -1], [-0.5, math.sqrt(3)/2, 0], [math.sqrt(3)/2, 0.5, 0] ]).T
q_C_A = np.array([3, 0, 2]).reshape(3, -1)

# T_C_A = np.eye(4)
# T_C_A[:3, :3] = R_C_A
# T_C_A[:3, 3] = q_C_A.flatten

T_C_A = np.block([
    [R_C_A, q_C_A], 
    [np.zeros(3), 1]
])

print(T_C_A)

# 现在求解T_A_C

T_A_C = np.block([
    [R_C_A.T, - R_C_A.T @ q_C_A],
    [np.zeros(3), 1]
])

print(T_A_C)

# 由此我们验证了刚体变换矩阵的逆矩阵关系
# T_C_A = (R_C_A, q_C_A)
#         (0,     1)
# T_A_C = T_C_A ^-1
#       = (R_C_A.T, -R_C_A.T @ q_C_A)
#         (0,     1)
```

```python
# 2.4
theta = 36.9
a = np.cos(np.deg2rad(theta))
b = np.sin(np.deg2rad(theta))
R_C_A = np.array([
    [a, b, 0],
    [b, -a, 0], 
    [0, 0, -1] 
]).T
q_C_A = np.array([-3, 4, 2]).reshape(3, -1)

T_C_A = np.block([
    [R_C_A, q_C_A], 
    [np.zeros(3), 1]
])
# print(T_C_A)
# T_A_C = np.block([
#     [R_C_A.T, - R_C_A.T @ q_C_A],
#     [np.zeros(3), 1]
# ])
# # print(T_A_C)

# 现在求T_A_B
R_B_A = np.array([
    [-1, 0, 0],
    [0, 0, -1], 
    [0, -1, 0] 
]).T
q_B_A = np.array([0, 4, 2]).reshape(3, -1)
T_A_B = np.block([
    [R_B_A.T, - R_B_A.T @ q_B_A],
    [np.zeros(3), 1]
])

T_C_B = T_A_B @ T_C_A
print(T_C_B)
# ok, 至此我们也验证了 刚体变换矩阵的链式法则
```
