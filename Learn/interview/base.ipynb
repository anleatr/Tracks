```python
# 1.写一个函数，返回 [1, 2, ..., n] 中所有偶数的平方，要求用一行列表推导式实现。
def get_odd_2(nums:list[int]):
    return [num*num  for num in nums if num %2 ==0]
nums = np.arange(1, 10, step=1) 
print(type(nums))
print(nums)

print(get_odd_2(nums))
```

```python
# 3.实现一个装饰器 @timer，用于打印被装饰函数的执行时间。
def time_decorator(func):
    def wrapper():
        start = time.time()
        func()
        print(f"use time:{time.time() - start}")
    return wrapper
@time_decorator
def simple_func():
    time.sleep(1)
# simple_func()
```

```python
# 4.手动实现一个简单的上下文管理器（如 with open_file("test.txt") as f:）。
with open("example.txt", 'r') as f:
    data = f.read()
# data

class open_file():
    def __init__(self, file_name, mode='r', encode="utf-8"):
        self.file_path = file_name
        self.mode = mode
        self.encode = encode
        self.file = None

    def __enter__(self):
        # 上下文管理器最重要的就是重写__enter__和__exit__方法，表示进入with和退出with时进行的操作
        print("enter")
        self.file = open(file=self.file_path, mode=self.mode, encoding=self.encode)
        return self.file
    
    def __exit__(self, exc_type, exc_val, exc_tb):
        print("exit")
        if self.file is not None:
            self.file.close()

g = open_file("example.txt")
with g as f:
    data = f.read()
data
    
```

```python
# 5.继承与抽象
from abc import ABC, abstractmethod
class Animal(ABC):
    def __init__(self):
        self.name:str
    
    @abstractmethod
    def speak(self):
        pass

class Dog(Animal):
    def __init__(self):
        super().__init__()
    def speak(self):
        self.name = "dog"
        print(f"{self.name} is barking")

class Cat(Animal):
    def __init__(self):
        super().__init__()
    def speak(self):
        self.name = "cat"
        print(f"{self.name} is miaomiaoing")
dog = Dog()
cat = Cat()
dog.speak()
cat.speak()
```

```python
# 文件操作

import re
# 单词计数
with open("example.txt", 'r') as f:
    data = f.read()

data = "".join(data.split())
data = re.sub('[^a-zA-Z]', '', data)
num_dict = {}
for char in data:
    if char in num_dict.keys():
        num_dict[char] += 1
    else:
        num_dict[char] = 1
num_dict
```

```python
# 目录 logs/ 下有多个 .log 文件（如 app1.log, app2.log）。
# 编写脚本将所有 .log 文件的内容按文件名排序后合并到 combined.log 中，每个文件内容前加上 === {filename} === 作为分隔。

import os
from pathlib import Path

file_path = "logs/"
files = os.listdir(file_path)
result = ""

for file_name in files:
    if file_name.split(".")[1] == 'log':
        with open(file_path+file_name, 'r') as f:
            data = f.read()
            data_struct =  f"\n=== {file_name} === \n" + data
            result += data_struct

with open(file_path+"combined.log", 'w', encoding='utf-8') as f:
    f.write(result)
    
```

```python
# 用生成器生成斐波那契数列
def gene_fbnq(max=1000):
    now = 0
    next = 1
    while True:
        if now < max:
            yield now
            tmp = next
            next += now
            now = tmp
        else:
            break
g = gene_fbnq()
for a in g:
    print(a)
```
