```python
# 把ab拼接，并且把第一个小写l转成大写
a = 'Hello, '
b = 'World!'
c = a + b
idx = c.find('l')

# c[idx] = 'L' # str是不可修改类型， 所有修改字符串的操作都是生成新的字符串

# 法一. 转list
d = list(c)
d[idx] = 'L'
final = "".join(d)
print(final)

# 法二. index和find， index是找不到就报错， find找不到返回-1
first_l_idx = c.index('l')
final = c[:first_l_idx] + 'L' + c[first_l_idx+1:]
print(final)

# 法三. replace
final = c.replace('l', 'L', 1) # (old, new, count)
print(final)
```

```python
# 我的丑陋方法
for idx, char in enumerate(c):
    if char == "l":
        c[idx] = char.upper()
        break
c
```

```python
# 排序专练
# 1. 对数组 c 先按第 2 个元素降序，再按第 1 个元素升序
c = [[4, 6], [2, 6], [5, 3], [1, 9], [3, 9]]
d = sorted(c, key=lambda x:x[1], reverse=True) # 按第二个降序
d = sorted(d, key=lambda x:x[0])
print(d)
e = sorted(c, key=lambda x: (-x[1], x[0]))
print(e)
```
