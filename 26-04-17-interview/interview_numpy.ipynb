```python
# 随机生成一个只有0到10整数，长度为20的数组
x = np.random.randint(0, 11, size=(20))
print(x)
```

```python
# 求数组 均值， 方差， 中位数， 众数
mean = np.mean(x)
max = np.max(x)
medi = np.median(x) # 前三种都有现成函数可以调
print(x)
print(f"mean:{mean}, max: {max}, medi: {medi}")

# 求众数
# 我的做法：用字典统计每个数字出现的次数
time_state = {}
for num in x:
    try:
        time_state[num] += 1
    except:
        time_state[num] = 1

max = sorted(time_state.values())[0]

zhongshu = []
for dict in time_state.items():
    if dict[1] == max:
        zhongshu.append(dict[0])
print(zhongshu)
zhongshu = np.mean(zhongshu)
print(f"众数:{zhongshu}")
```
