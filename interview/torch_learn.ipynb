```python
# 张量创建与操作
import torch

# 1. 创建一个 3×3 的随机张量，并将其第 2 行设为 0。
a = torch.rand(3, 3)
a[1][:] = 0
a
```

```python
# 2. 将 NumPy 数组 np.array([1,2,3]) 转换为 PyTorch 张量，要求梯度可追踪。
import numpy as np
a = np.array([1, 2, 3])
b = torch.tensor(a, dtype=torch.float32, requires_grad=True)
c = torch.tensor(a)
b.dtype, c.dtype
```

```python
# 3. 给定 x = torch.tensor(2.0, requires_grad=True)，计算 y = x**3 + 2*x 并求 dy/dx 在 x=2 处的值。
x = torch.tensor(2.0, requires_grad=True)
y = x**3 + 2*x
y.backward()
# torch和numpy最大区别就是可以计算梯度吗？
x.grad
```

```python
# 4. 如何将张量移到 GPU？如何确保模型和数据在同一设备上？
# 两种方法，一种直接创建在gpu上，一种通过to移动
# 必须要a = a.to(device)不能直接a.to(device)
a = torch.randn(3, 4)
device = "cuda" if torch.cuda.is_available() else "cpu"
a = a.to(device)
a.device
```
