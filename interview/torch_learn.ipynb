```python
# 实现一个图像分类器，经过卷积，池化，最后输出图像类别

batch, channel , w, h = 1, 3, 256, 256
# input = torch.randint(0, 255, size=(batch, channel, w, h))
input = torch.randn((batch, channel, w, h), dtype=torch.float32)

class SimpleClassify(nn.Module):
    def __init__(self, ):
        super().__init__()

        
        ## 第一层卷积
        self.conv1 = nn.Conv2d(in_channels=3, out_channels=6, kernel_size=5, padding=2, stride=2) # [6, 128, 128]
        self.pool1 = nn.AvgPool2d(2) # [6, 64, 64], 默认stride=kernel_size

        self.fc1 = nn.Linear(64*64*6, 756)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(756, 10)

    def forward(self, x):
        x = self.conv1(x)
        x = self.relu(x)

        x = self.pool1(x) # [1, 6, 64, 64]
        print(self.pool1.stride)
        x = x.flatten(start_dim=1) # 不能把batch也压平
     
        x = self.fc1(x)

        x = self.relu(x)

        return self.fc2(x)
        

model = SimpleClassify()
y = model(input)
logit = torch.softmax(y, dim=1)
print(logit.shape)

```
