```python
import numpy as np

# 1. 数组创建与索引
a = np.eye(5)
b = np.random.randn(5,5)
c = b[np.where(b>0.5)]
b[np.where(b<=0.5)] = 0
b,c
```

```python
# 2.广播与计算
a = np.random.randn(3,4)
b = np.random.randn(4,)

# a.shape, b.shape
c = a + b
d = a * b
a,b, c, b.shape
# 为什么(3,4)和(4,)可以广播， 但是(3,4)和(3,)不能广播？
# 广播的规则遵循从右往左对其，前者：4和4对其，补上3，后者：4无法和3对其
# 但是1是特殊的，比如(3,4)可以和(3,1)广播 
```

```python
# 3.L2归一化
a = np.random.randn(3,4)
print(a)
l2_norm = np.linalg.norm(a, axis=1) # 线性代数库
print(l2_norm)
a = a/l2_norm.reshape(3,1)
print(a)
print( np.linalg.norm(a, axis=1))
```
