```python
# 2.广播与计算
import numpy as np
a = np.random.randn(3,4)
b = np.random.randn(4,)

# a.shape, b.shape
c = a + b
d = a * b
e = a @ b
c.shape, d.shape, e.shape
# 为什么(3,4)和(4,)可以广播， 但是(3,4)和(3,)不能广播？
# 广播的规则遵循从右往左对其，前者：4和4对其，补上3，后者：4无法和3对其
# 但是1是特殊的，比如(3,4)可以和(3,1)广播 
```

```python
# 4. 对比 Python 循环和 NumPy 向量化操作在求和上的速度差异
# 什么是向量化操作？ 比如我们举个求和的例子

import time
# python 循环
start = time.time()
total = 0
for i in range(1500000):
    total += i
print(total) # 1124999250000
print(f"use_time:{time.time() - start}")

# 向量化操作
start = time.time()
sum = np.sum(np.arange(0,1500000, dtype=np.int64))
print(sum) # -282181552， 默认为int32， 最大值为 2^31
print(f"use_time:{time.time() - start}")

# numpy的整数溢出了
```
